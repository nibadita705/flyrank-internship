## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

* *Total analyzed URLs in starter dataset:* 30,000 pages
* *Decaying / Declining pages identified:* 16,260 pages (54.2% of the total dataset)
* *Traffic impact:* Median 90-day impressions on decaying URLs is ~1,850, representing significant reclaimable organic reach.


---



```python
import os
import subprocess
import pandas as pd

# Clone repository if running in a fresh Colab environment without local data
csv_path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(csv_path):
    if not os.path.exists("flyrank-internship"):
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/nibadita705/flyrank-internship"], check=True)
    if os.path.exists("flyrank-internship/data/raw/content_refresh_anonymized.csv"):
        os.chdir("flyrank-internship")

# 1. Load the starter CSV dataset
df = pd.read_csv(csv_path)

# 2. Compute key real numbers demonstrating business scope and urgency
total_urls = len(df)
declining_urls = (df["trend_direction"].str.lower() == "down").sum()
declining_percentage = (declining_urls / total_urls) * 100
median_impressions = df[df["trend_direction"].str.lower() == "down"]["impressions_90d"].median()

# 3. Print verified metrics
print("--- Starter Dataset Real Numbers ---")
print(f"1. Total URLs Analyzed: {total_urls:,}")
print(f"2. Decaying / Declining Pages: {declining_urls:,} ({declining_percentage:.1f}% of total)")
print(f"3. Median 90-Day Impressions on Decaying URLs: {median_impressions:,.0f}")
```
